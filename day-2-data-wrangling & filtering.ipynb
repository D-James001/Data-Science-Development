```python
# Filtering Anomalies
# Dataset: Nigerian Economics Data
# Scenario: The governments wants to calculate the average income of working-age adults.
# Reason: They need to set a new minimum wage
```

```python
# Package of interest
import pandas as pd
```

```python
# Load our dataset into a dataframe
economic_data = pd.read_csv('/kaggle/input/datasets/awofiranyejames/nigerian-economic-data/nigeria_economic_data.csv')
economic_data.head(25)
```

```python
# Scope & Focus
# Remove logical impossibilities
# Targeted slicing
# We must create a new dataset, but only keep rows where Age is 18 or older
adult_only_eco_data = economic_data[(economic_data['Age'] >= 18) & (economic_data['Age'] <= 65)]
```

```python
adult_only_eco_data.head()
#adult_only_eco_data.shape
```

```python
print(f"Original Dataset Size:", len(economic_data))
print(f"New Dataset Size:", len(adult_only_eco_data))
```

*Scenario:

- You are a non-profit organization

```python
# Dataset of interest
health_data = pd.read_csv('/kaggle/input/datasets/awofiranyejames/nigerian-nursing-mother-health-care/nigeria_nursing_mothers_healthcare.csv')
health_data.head(3)
```

```python
health_data['Wealth_Quintile'].unique()
```

```python
# We need to create a new dataset from the existing one
# Poorest & Rural
targeted_mothers = health_data[(health_data['Residence'] == 'Rural') & (health_data['Wealth_Quintile'] == 'Poorest')]
```

```python
print(f"Original Dataset Size:", len(health_data))
print(f"New Dataset Size:", len(targeted_mothers))
```

**Investigating Suspicious Data

*Dataset: Nigerian Unemployment Data

*Scenario: Find suspicious data that needs human review

```python
# Dataset of interest
unemployment_data = pd.read_csv('/kaggle/input/datasets/awofiranyejames/nigeria-unemployment-dataset/nigeria_unemployment_missing_data.csv')
unemployment_data.head(2)
```


1. Focus: Age and Educational level
2. Is it impossible to have a university degree at 16? No, but it is highly unusual
3. Let's filter to find anyone under 18 who claims to have a university degree


```python
unemployment_data['Education_Level'].unique()
```

```python
suspicious_data = unemployment_data[(unemployment_data['Age'] < 18) & (unemployment_data['Education_Level'] == 'Tertiary')]
```

```python
print(f"Number of Suspicious Records Found:", len(suspicious_data))
```
