```python
# Package of Interest
import pandas as pd
```

```python
# Inspection of Unemployment Data
unemployment_data = pd.read_csv('/kaggle/input/datasets/awofiranyejames/unemployment-dataset/nigeria_unemployment_missing_data.csv')
unemployment_data.head()
```

```python
# Inspect the ingredients
# You are looking for rot (Missing values)
# Why are you looking out for rot (Missing Values)?
# Find out how many blank spaces are in your data
unemployment_data.isnull().sum()
```

```python
# What are the implications for the missing values?
unemployment_data.shape
```

```python
# We must approach wrangling with several assumption
# 1. We may not need the missing values
# 2. We must understand why the missing value exist
# 3. We must have contingencies for dealing with missing data
```

```python
# We may not need the missing values
unemployment_data_1 = unemployment_data.copy()
```

```python
unemployment_data_1.head()
```

```python
# Take away all the missing values
unemployment_data_1.dropna(inplace=True)
```

```python
unemployment_data.shape
```

```python
# Dataframe without missing value
unemployment_data_1.shape
```

```python
# Fix the years of experience column
# We think "Years of Experience" is subjective
# We will fill the missing values with the average years of experience
# Calculate the average years of experience
average_experience = unemployment_data['Years_Of_Experience'].mean()
```

```python
average_experience
```

```python
# Before we fix the column with missing values
unemployment_data_avg_years = unemployment_data.copy()
```

```python
# The Years of Experience has been filled with the average years of experience
unemployment_data_avg_years['Years_Of_Experience'].fillna(average_experience, inplace=True)
```

```python
unemployment_data_avg_years.isnull().sum()
```

```python
# Tracking Education_Level in Unemployment Data (Average Years)
unemployment_data_avg_years.Education_Level.value_counts(normalize=True)
```

```python
unemployment_data_avg_years['Education_Level'] = unemployment_data_avg_years['Education_Level'].replace(float('nan'), None)
print(unemployment_data_avg_years['Education_Level'].isnull().sum()) # Should be 0 now

unemployment_data_avg_years['Education_Level'].head()
```

```python
unemployment_data_avg_years.Education_Level.value_counts()
```

```python
unemployment_data_avg_years['Education_Level'] = unemployment_data_avg_years['Education_Level'].fillna('Unknown')
unemployment_data_avg_years['Education_Level'].value_counts()
```
