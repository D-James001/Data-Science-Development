**Introduction to Data Governance**

- The Big Difference: Data Governance vs. Data Management

- The Pillars of Data Governance

- Data Wrangling: Chopping, Cleaning, and Peeling your ingredients

- Machine Learning: Using a high-tech oven

- Data Governance: It is the restaurants food safety code

- Data Governance prevents you from misbehaving with data

*The Big Difference*

- Data Management: Writing Python code, Cleaning data, building a pipeline, exploring the data

- Data Governance: This is a strategic framework.

- With Data Governance: We define who is allowed to touch data, what that data can be used for, how the data quality is monitored, and how long are you allowed to keep the data before throwing it away.

*The Six Pillars of Data Governance*

- Accountability: The designated owner or data steward

- Transparency: You must clearly explain to your customers and regulators what data you are holding, where it came from, how it flows through your pipeline...

- Data Quality: Standards for accuracy, consistency & completeness.

- Security: This controls who has specific access to data.

- Purpose Limitation: Using data for a specific purpose. If it must be repurposed, there must be explicit consent granted.

- Data Retention: You cannot keep customer data forever. Governance defines strict expiration date for data categories.

**Introduction to Inferential Statistics**

- A spoonful of soup is used to make decision about the entire pot of soup.

*Descriptive Statistics:* Simply stating the facts about the data in front of you

*Inferential Statistics:* It is the science using small sample data to draw reliable and mathematically backed Conclusion about the much larger population.


*The median age of the Nigerian population is approximately 18.3 years*

*The median age of the United States population is 39.4 years old*

**Why is Inferential Statistics the Backbone of Machine Learning?**

- Feature Selection: If you have 50 columns in a dataframe, which one will affect your desired outcome, and which ones will be useless? Inferential tests

- A/B Testing: It is a standard for verifying differences between two activities/situations/experiences. Inferential statistics gives us the math to prove relevance.

- Model Validation: Did the 1% improvement of a model mean that the code was smarter or the computer got lucky? Hypothesis testing is used to confirm if performance is genuine or due to luck.

**Four Pillars**

- The Population vs. The Sample: The population is the entire group you want to study. The Sample is a smaller slice of the population

- The Null Hypothesis(H0) vs. The Alternative Hypothesis(Ha): "A defendant is innocent until proven guilty", "The defendant is guilty and it can be proven beyond a reasonable doubt"

- The p-value: The p-value is a number between 0 and 1 that measures the probability that a result is a complete fluke. "If the p-value is low(usually less than 0.05), the null hypothesis must go.

- Confidence Intervals: This a mathematical net designed to capture the true population average with a level of certainty.

**The Independent T-Test**

*We are comparing two independent averages.*

- "Are the average values of two independent groups different enough that we should believe the difference is real?"

- We are comparing two separate groups: people, customers, patients, students, etc

*Scenario:* The company launches a new training program

- They want to know: Do people who receive the new training score differently from people who receive the old training?

- Group A -> Old Training

- Group B -> New Training

*Stats for old & new training:*

- Old Training: 65, 70, 72, 68, 75, 71, 69

- New Training: 78, 82, 80, 85, 79, 83, 81

*The Data Science Question:*

- Is the difference large enough for HR to believe the new training actually make a difference?

*This is where the Independent T-Test comes in*

- Different People are tested - Independent t-test
- Same People are tested twice - paired t-test

```python
# Simulate a realistic example
# We are testing whether a new marketing campaign increases customer spending

np.random.seed(42)
```

```python
control = np.random.normal(loc=500, scale=100, size=100)
treatment = np.random.normal(loc=550, scale=100, size=100)
# Control -> Customers who did not receive the campaign
# Treatment -> Customers who received the campaign
```

```python
control.mean()
```

```python
treatment.mean()
```

```python
# Control: 506
# Treatment: 560
```

```python
# The Difference
difference = treatment.mean() - control.mean()
```

```python
difference
```

```python
# The treament group spent approximately $54 more than the control group
# Is $54 enough evidence that the campaign really works?
```

```python
# Run an Independent T-Test
from scipy.stats import ttest_ind
```

```python
t_stat, p_value = ttest_ind(treatment, control)
```

```python
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
# The Incorrect Analysis
# "There is a 0.01% probability that the null hypothesis is True"

# The Correct Analysis
# If there really were no difference between the two groups,
# getting a difference this large would be unusual

# There is a strong evidence that the groups differ
# The campaign has a significant difference in customer spending
```

```python
# The US Educational Data
# Do the Public & Private Non-Profit Schools have different graduation rates
```

```python
us_edu_data['control'].unique()
```

```python
# Null Hypothesis: There isn't any difference in their graduate rates
```

```python
public_grads = us_edu_data[us_edu_data['control'] == 'Public']['graduation_rate_pct']
```

```python
private_grads = us_edu_data[us_edu_data['control'] == 'Private Non-Profit']['graduation_rate_pct']
```

```python
# Run the Independent T-Test
t_stat, p_value = ttest_ind(public_grads, private_grads)
```

```python
print(f"Calculated T-Statistics: {t_stat:.4f}")
print(f"Calculated p-value: {p_value:.4f}")
```

```python
# Interpret the result for our stakeholder
alpha = 0.05
```

```python
if p_value < alpha:
  print("Result: Significant! We reject the Null Hypothesis. There is a real difference.")
else:
  print("Result: Not significant. We fail to reject the Null. Any difference was just a fluke.")
```

**One-Way ANOVA - Comparing Multiple Groups**

*The Relatable Analogy: The Tomato Factory*

- Testing three brands of tomato fertilizer

- Brand A, Brand B, & Brand C - Which one makes the plants grow the tallest.

- Why don't we run three separate t-tests?

- Statistical Exhaustion - Run three t-tests back-to-back, our overall chance of making mistakes increases.

*The Analysis of Variance:* It checks all the groups simultaneously. Is there any statistically significant differences among these averages, or are they the same?

```python
# Data
us_edu_data.head(3)
```

```python
# Is there a significant difference in the average student-teacher ratio
# across different geographical locales
```

```python
# Unique locales
us_edu_data['nces_locale'].unique()
```

```python
# Group our student-teacher ratios by Geographic locale
city_ratios = us_edu_data[us_edu_data['nces_locale'] == 'City - Large']['student_teacher_ratio']
suburban_ratios = us_edu_data[us_edu_data['nces_locale'] == 'Suburban - Large']['student_teacher_ratio']
rural_ratios = us_edu_data[us_edu_data['nces_locale'] == 'Rural - Fringe']['student_teacher_ratio']
```

```python
# Run the One-Way ANOVA Test
f_stat, p_value = stats.f_oneway(city_ratios, suburban_ratios, rural_ratios)
```

```python
# H0 - Staffing ratios differ significantly across locales
# Ha - Any difference in ratio is just a random fluke
```

```python
# Interpret Our Result
alpha = 0.05
if p_value < alpha:
  print("Result: Significant! We reject the Null Hypothesis. Staffing ratios differ significantly across locales.")
else:
  print("Result: Not significant. We fail to reject the Null. Any difference was just a fluke.")
```

**The Chi-Square Test - Connecting Categories**

*The Relatable Analogy: The Movie Theatre Snack Bar*

- Does a customer's choice of snack (popcorn, Fishpie, or Donuts) depend on the movie they are watching (Action vs. Comedy)?

- Snack choice and the movie type

- A Chi-Square Test of Independence: Whether the snack choice is independent of the movie type, or if there is a strong, non-random connection

```python
# Real-World Use Case
# Does a school's `title_i_status`have a structural association with its STEM certification status
# `title_i_status`| `stem_certified_flag`
```

```python
us_edu_data['stem_certified_flag'].unique()
```

```python
us_edu_data['title_i_status'].unique()
```

```python
# Create a contingency table with counts of the categories
contingency_table = pd.crosstab(us_edu_data['title_i_status'], us_edu_data['stem_certified_flag'])
```

```python
contingency_table
```

```python
# Run the Chi-Test of Independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)
```

```python
print(p_value)
```

```python
# Ho - STEM certification is structurally associated with Title I eligibility
# Ha - STEM certification is independent of Title I funding status.
# Interpret
alpha = 0.05
if p_value < alpha:
  print("Result: Significant! We reject the Null Hypothesis. STEM certification is associated with Title I eligibility.")
else:
  print("Result: Not significant. We fail to reject the Null.")
```

**Correlation Coefficient Significance - The Smoking Gun**

*The Relatable Analogy: Height vs. Weight*

- The correlation coefficient(r) elegantly condenses the complex scatterplot into a single number between -1 & 1

- What if we calculate a correlation of r =.4 on a tiny sample of data? Is that real? or is it just happenstance?

- We must calculate the standard error or our correlation to test the Null Hypothesis

```python
# `per_pupil_expenditure_usd`
# `graduation_rate_pct`
# Does the school's per_pupil_expenditure_usd correlate with its graduation_rate_pct
# Testing the significance of the correlation coefficient
```

```python
grad_rate = us_edu_data['graduation_rate_pct']
spending = us_edu_data['per_pupil_expenditure_usd']
```

```python
grad_rate.head()
```

```python
spending.head()
```

```python
# Pearson Correlation Coefficient
r_coeff, p_value = stats.pearsonr(grad_rate, spending)
```

```python
print(f"Calculated Correlation Coefficient: {r_coeff:.4f}")
print(f"Calculated p-value: {p_value:.4f}")
```

```python
# Interpret
alpha = 0.05
if p_value < alpha:
  print("Result: Significant! There is a real correlation.")
else:
  print("Result: Not significant. Any correlation was just a coincidence.")
```
